```python
"""!!! ПОЛНЫЙ КОД ВСЕГО ВЫШЕ !!!"""

from sympy import Symbol, S, solveset, Eq
from sympy.calculus.util import continuous_domain, function_range

x = Symbol("x")

# =========================================================
# ШАГ 0: Задаем функцию
# =========================================================
f = (x**3) / (x**2 - 1)
print("--- ИССЛЕДУЕМ f(x) = x^3 / (x^2 - 1) ---")

# =========================================================
# ШАГ 1: Сама функция
# =========================================================
print("\n[1] САМА ФУНКЦИЯ:")
print("Область определения:", continuous_domain(f, x, S.Reals))
print("Область значений:", function_range(f, x, S.Reals))

# =========================================================
# ШАГ 2: Первая производная
# =========================================================
print("\n[2] ПЕРВАЯ ПРОИЗВОДНАЯ (скорость изменения):")
f_prime = f.diff(x)
print("Формула f'(x):", f_prime)

# ВАЖНО: Проверяем область определения производной!
print("Область определения f'(x):", continuous_domain(f_prime, x, S.Reals))

# Ищем корни (где f'(x) = 0)
roots_d1 = solveset(Eq(f_prime, 0), x, domain=S.Reals)
print("Корни f'(x) = 0:", roots_d1)

# =========================================================
# ШАГ 3: Вторая производная
# =========================================================
print("\n[3] ВТОРАЯ ПРОИЗВОДНАЯ (кривизна/изгибы):")
f_double_prime = f.diff(x, 2)
print("Формула f''(x):", f_double_prime)

# ВАЖНО: Проверяем область определения второй производной!
print("Область определения f''(x):", continuous_domain(f_double_prime, x, S.Reals))

# Ищем корни (где f''(x) = 0)
roots_d2 = solveset(Eq(f_double_prime, 0), x, domain=S.Reals)
print("Корни f''(x) = 0:", roots_d2)
```

```python
"""!!! ПОЛНЫЙ КОД ФИНАЛЬНЫЙ  !!!"""

# 1. Импортируем нужные инструменты
from sympy import Symbol, S, solveset, Eq, limit, oo
from sympy.calculus.util import continuous_domain, function_range
from sympy import Symbol, diff, factor, together

# 2. Создаем переменную x
x = Symbol("x")

# 3. Задаем нашу функцию y
y = (x**3) / (2*(x + 5)**2)
print("=== ИССЛЕДОВАНИЕ ФУНКЦИИ ===")

# ==========================================
# БЛОК 1: Сама функция y
# ==========================================
print("\n[1] САМА ФУНКЦИЯ y:")
# Ищем область определения
domain_y = continuous_domain(y, x, S.Reals)
print("Область определения y:", domain_y)
# Ищем область значений
range_y = function_range(y, x, S.Reals)
print("Область значений y:", range_y)

# ==========================================
# БЛОК 2: Первая производная y_1
# ==========================================
print("\n[2] ПЕРВАЯ ПРОИЗВОДНАЯ y_1:")
# Находим первую производную
y_1 = y.diff(x)
y_1_beautiful = factor(together(y_1))
print("Формула y_1:", y_1_beautiful)

# Ищем область определения y_1
domain_y_1 = continuous_domain(y_1, x, S.Reals)
print("Область определения y_1:", domain_y_1)
# Ищем корни y_1 (где производная равна 0)
roots_y_1 = solveset(Eq(y_1, 0), x, domain=S.Reals)
print("Корни y_1 = 0:", roots_y_1)

# ==========================================
# БЛОК 3: Вторая производная y_2
# ==========================================
print("\n[3] ВТОРАЯ ПРОИЗВОДНАЯ y_2:")
# Находим вторую производную
y_2 = y.diff(x, 2)
y_2_beautiful = factor(together(y_2))
print("Формула y_2:", y_2_beautiful)
# Ищем область определения y_2
domain_y_2 = continuous_domain(y_2, x, S.Reals)
print("Область определения y_2:", domain_y_2)
# Ищем корни y_2
roots_y_2 = solveset(Eq(y_2, 0), x, domain=S.Reals)
print("Корни y_2 = 0:", roots_y_2)

# подставляем корни первой во вторую производную

result = y_2.subs(x, -15)
print("Значение второй производной в точке экстремума первой", result)

# ==========================================
# БЛОК 4: Асимптоты
# ==========================================
print("\n[4] АСИМПТОТЫ:")
# Ищем вертикальные (в точках разрыва 1 и -1)
limit_right = limit(y, x, 1)
print("Предел при x -> 1:", limit_right)
limit_left = limit(y, x, -1)
print("Предел при x -> -1:", limit_left)

# Ищем горизонтальную (на бесконечности)
limit_inf = limit(y, x, oo)
print("Предел при x -> бесконечности:", limit_inf)

# Ищем наклонную (если нет горизонтальной)
if limit_inf == oo or limit_inf == -oo:
    # Считаем наклон k
    k = limit(y / x, x, oo)
    print("Наклон k:", k)
    # Считаем сдвиг b
    b = limit(y - k*x, x, oo)
    print("Сдвиг b:", b)
    # Выводим итоговое уравнение
    print(f"Уравнение наклонной асимптоты: y = {k}*x + {b}")
else:
    print("Наклонной асимптоты нет.")
```

```python
"""!!! ПОЛНЫЙ КОД ФИНАЛЬНЫЙ С ИССЛЕДОВАНИЕМ НА ЧЕТНОСТЬ/НЕЧЕТНОСТЬ !!!"""

from sympy import Symbol, S, solveset, Eq, limit, oo, simplify
from sympy.calculus.util import continuous_domain, function_range

x = Symbol("x")
y = (x**3) / (2*(x + 5)**2)
print("=== ПОЛНЫЙ АНАЛИЗ ФУНКЦИИ y = x^3 / (2*(x + 5)^2) ===")

# [1] ЧЕТНОСТЬ / НЕЧЕТНОСТЬ
print("\n[1] ЧЕТНОСТЬ / НЕЧЕТНОСТЬ:")
y_neg_x = y.subs(x, -x)
if simplify(y_neg_x - y) == 0:
    print("Функция ЧЕТНАЯ")
elif simplify(y_neg_x + y) == 0:
    print("Функция НЕЧЕТНАЯ")
else:
    print("Функция НИ ЧЕТНАЯ, НИ НЕЧЕТНАЯ")

# [2] ПЕРЕСЕЧЕНИЯ С ОСЯМИ
print("\n[2] ПЕРЕСЕЧЕНИЯ С ОСЯМИ:")
print("Пересечение с осью Y (x=0):", y.subs(x, 0))
print("Пересечение с осью X (y=0):", solveset(Eq(y, 0), x, domain=S.Reals))

# [3] ОБЛАСТЬ ОПРЕДЕЛЕНИЯ И ЗНАЧЕНИЙ
print("\n[3] ОБЛАСТЬ ОПРЕДЕЛЕНИЯ И ЗНАЧЕНИЙ:")
print("Область определения y:", continuous_domain(y, x, S.Reals))
print("Область значений y:", function_range(y, x, S.Reals))

# [4] ПЕРВАЯ ПРОИЗВОДНАЯ
print("\n[4] ПЕРВАЯ ПРОИЗВОДНАЯ (y_1):")
y_1 = y.diff(x)
y_1_beautiful = factor(together(y_1))
print("Формула y_1:", y_1_beautiful)
print("Область определения y_1:", continuous_domain(y_1, x, S.Reals))
print("Корни y_1 = 0 (экстремумы):", solveset(Eq(y_1, 0), x, domain=S.Reals))

# [5] ВТОРАЯ ПРОИЗВОДНАЯ
print("\n[5] ВТОРАЯ ПРОИЗВОДНАЯ (y_2):")
y_2 = y.diff(x, 2)
y_2_beautiful = factor(together(y_2))
print("Формула y_2:", y_2_beautiful)
print("Область определения y_2:", continuous_domain(y_2, x, S.Reals))
print("Корни y_2 = 0 (точки перегиба):", solveset(Eq(y_2, 0), x, domain=S.Reals))

# [6] АСИМПТОТЫ
print("\n[6] АСИМПТОТЫ:")
print("Предел при x -> -5:", limit(y, x, -5))
limit_inf = limit(y, x, oo)
print("Предел при x -> oo:", limit_inf)
if limit_inf == oo or limit_inf == -oo:
    k = limit(y / x, x, oo)
    b = limit(y - k*x, x, oo)
    print(f"Наклонная асимптота: y = {k}*x + ({b})")
```

```python


"""!!! ПОЛНЫЙ КОД ФИНАЛЬНЫЙ С ИССЛЕДОВАНИЕМ НА ЧЕТНОСТЬ/НЕЧЕТНОСТЬ И ФИНАЛЬНЫМ ВЫВОДОМ!!!"""

python
from sympy import Symbol, S, solveset, Eq, limit, oo, simplify
from sympy.calculus.util import continuous_domain, function_range

# 1. Переменная и функция
x = Symbol("x")
y = (x**3) / (2*(x + 5)**2)

print("="*60)
print("АНАЛИЗ ФУНКЦИИ: y = x^3 / (2*(x + 5)^2)")
print("="*60)

# ==========================================
# БЛОК 1: Четность/Нечетность
# ==========================================
print("\n[1] ЧЕТНОСТЬ / НЕЧЕТНОСТЬ:")
y_neg_x = y.subs(x, -x)
if simplify(y_neg_x - y) == 0:
    print("Функция ЧЕТНАЯ (симметрична оси Y).")
elif simplify(y_neg_x + y) == 0:
    print("Функция НЕЧЕТНАЯ (симметрична центру).")
else:
    print("Функция общего вида (ни четная, ни нечетная).")

# ==========================================
# БЛОК 2: Пересечения с осями
# ==========================================
print("\n[2] ПЕРЕСЕЧЕНИЯ С ОСЯМИ:")
y_intercept = y.subs(x, 0)
print(f"С осью Y (при x=0): y = {y_intercept}")
x_intercepts = solveset(Eq(y, 0), x, domain=S.Reals)
print(f"С осью X (при y=0): x = {x_intercepts} (то есть точка (0,0))")

# ==========================================
# БЛОК 3: Области определения и значений
# ==========================================
print("\n[3] ОБЛАСТЬ ОПРЕДЕЛЕНИЯ И ЗНАЧЕНИЙ:")
domain_y = continuous_domain(y, x, S.Reals)
print(f"Область определения: {domain_y}")
print("   Перевод: Все числа от минус бесконечности до плюс бесконечности, КРОМЕ x = -5.")
print("   Почему: Если подставить -5 в знаменатель, получится 0, а на ноль делить нельзя.")

range_y = function_range(y, x, S.Reals)
print(f"Область значений: {range_y}")
print("   Перевод: Функция может выдавать любые числа (от -oo до +oo).")

# ==========================================
# БЛОК 4: Первая производная
# ==========================================
print("\n[4] ПЕРВАЯ ПРОИЗВОДНАЯ (Экстремумы):")
y_1 = y.diff(x)
y_1_simplified = simplify(y_1) # Упрощаем формулу для читаемости
print(f"Формула y_1: {y_1_simplified}")
print("   (Если перевести в обычную дробь: y' = x^2 * (x + 15) / (2 * (x + 5)^3))")

domain_y_1 = continuous_domain(y_1, x, S.Reals)
print(f"Область определения y_1: {domain_y_1}")
print("   Перевод: Все числа, кроме x = -5 (там разрыв).")

roots_y_1 = solveset(Eq(y_1, 0), x, domain=S.Reals)
print(f"Корни y_1 = 0: {roots_y_1}")
print("   Перевод: В точках x = 0 и x = -15 график замирает (это горы или впадины).")

# ==========================================
# БЛОК 5: Вторая производная
# ==========================================
print("\n[5] ВТОРАЯ ПРОИЗВОДНАЯ (Точки перегиба):")
y_2 = y.diff(x, 2)
y_2_simplified = simplify(y_2)
print(f"Формула y_2 (упрощенная): {y_2_simplified}")

domain_y_2 = continuous_domain(y_2, x, S.Reals)
print(f"Область определения y_2: {domain_y_2}")

roots_y_2 = solveset(Eq(y_2, 0), x, domain=S.Reals)
print(f"Корни y_2 = 0: {roots_y_2}")
print("   Перевод: В точке x = 0 график меняет форму (с чаши на зонтик).")

# ==========================================
# БЛОК 6: Асимптоты
# ==========================================
print("\n[6] АСИМПТОТЫ:")
# Проверяем точку разрыва -5
limit_break = limit(y, x, -5)
print(f"Предел при x -> -5: {limit_break}")
if limit_break == oo or limit_break == -oo:
    print("   Перевод: Здесь ВЕРТИКАЛЬНАЯ асимптота. График улетает в бесконечность.")

# Проверяем на бесконечности
limit_inf = limit(y, x, oo)
print(f"Предел при x -> oo: {limit_inf}")
if limit_inf == oo or limit_inf == -oo:
    print("   Перевод: Горизонтальной асимптоты нет (график улетает вверх).")
    print("   Ищем наклонную...")
    k = limit(y / x, x, oo)
    b = limit(y - k*x, x, oo)
    print(f"   Наклонная асимптота: y = {k}*x + ({b})")
    print("   Перевод: При больших x график прижимается к прямой линии y = 0.5x - 5.")
print("\n" + "="*60)
print("АНАЛИЗ ЗАВЕРШЕН. ИНСТРУМЕНТ ДЛЯ ML ГОТОВ!")
print("="*60)
```
