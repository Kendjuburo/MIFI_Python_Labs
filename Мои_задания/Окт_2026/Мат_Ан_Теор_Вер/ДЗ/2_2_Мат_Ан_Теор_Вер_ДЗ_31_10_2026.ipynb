## Частные производные ##

```python
from sympy import symbols, cos, diff

a, b, c = symbols('a b c', real=True)
f = 5*a*b - a*cos(c) + a**8 + c**2*b

print(diff(f, a))
```

## Частные производные в точках ##

```python
from sympy import symbols, cos, diff

# 1. Определяем переменные
a, b, c = symbols('a b c', real=True)

# 2. Задаем функцию
f = 5*a*b - a*cos(c) + a**8 + c**2*b

# 3. Находим частные производные (символьные формулы)
df_da = diff(f, a)
df_db = diff(f, b)
df_dc = diff(f, c)

# 4. Задаем точку, в которой хотим посчитать значения (например, a=1, b=1, c=1)
# Используем словарь: {переменная: значение}
point = {a: 1, b: 1, c: 1}

# 5. Подставляем точку в каждую производную
val_a = df_da.subs(point)
val_b = df_db.subs(point)
val_c = df_dc.subs(point)

# 6. Выводим результаты
print("Частная производная по a:", df_da)
print("Значение в точке (1,1,1):", val_a)
print("-" * 30)

print("Частная производная по b:", df_db)
print("Значение в точке (1,1,1):", val_b)
print("-" * 30)

print("Частная производная по c:", df_dc)
print("Значение в точке (1,1,1):", val_c)
```
