### NA Градинетный спуск - РАССЧЕТ ДЛЯ 3-ч переменных ###

```python
from sympy import symbols, diff
import numpy as np

# 1. Задаем символы
x, y, z = symbols('x y z')

# 2. Задаем функцию
f = 2*x**2 - 4*x*z + 4*y**2 - 8*y*z + 9*z**2 + 4*x + 8*y - 20*z

# 3. Находим градиент (частные производные)
df_dx = diff(f, x)
df_dy = diff(f, y)
df_dz = diff(f, z)

# 4. Задаем текущую точку и alpha
point = {x: 0, y: 0, z: 0}
alpha = 0.25

# 5. Подставляем точку в производные (получаем вектор градиента)
grad_vec = np.array([
    df_dx.subs(point),
    df_dy.subs(point),
    df_dz.subs(point)
], dtype=float)

print("Градиент в точке (0,0,0):", grad_vec)

# 6. Делаем шаг градиентного спуска
current_point = np.array([0.0, 0.0, 0.0])
new_point = current_point - alpha * grad_vec

print("Новая точка:", new_point)
```

###  NA Градинетный спуск - РАССЧЕТ ДЛЯ 2-х переменных вручную ###

```python
import numpy as np

# 1. Задаем функцию
def f(x, y):
    return 2*x**2 - 4*x*y + y**4 + 2

# 2. Задаем градиент (частные производные)
def grad(x, y):
    df_dx = 4*x - 4*y
    df_dy = -4*x + 4*y**3
    return np.array([df_dx, df_dy])

# 3. Градиентный спуск
def gradient_descent(start_point, lr=0.01, epochs=1000):
    point = np.array(start_point, dtype=float)
    for _ in range(epochs):
        g = grad(point[0], point[1])
        point = point - lr * g
    return point

# Запускаем из точки (2, 2) - подальше от седла (0,0)
min_point = gradient_descent([2.0, 2.0])
print(f"Точка минимума: x={min_point[0]:.4f}, y={min_point[1]:.4f}")
print(f"Значение функции в этой точке: {f(min_point[0], min_point[1]):.4f}")
```

###  Градинетный спуск - УНИВЕРСАЛЬНЫЙ!!!! ###


```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# ==============================================================================
# БЛОК 1: ВВОД ПЕРЕМЕННЫХ И ФОРМУЛЫ
# ==============================================================================
# 1.1. Задаем переменные (3 штуки)
x, y, z = sp.symbols('x y z')

# 1.2. Задаем ФОРМУЛУ (внимательно со знаками)
formula = x**3 - 2*(x**2) + y**2 + z**2 - 2*x*y + x*z - y*z + 3*z

# ==============================================================================
# БЛОК 2: АВТОМАТИЧЕСКИЙ РАСЧЕТ ПРОИЗВОДНЫХ (ИСПРАВЛЕНО)
# ==============================================================================
# Собираем ВСЕ переменные в список
vars_list = [x, y, z]

# SymPy сам находит производные по каждой переменной из списка
derivatives = [sp.diff(formula, var) for var in vars_list]

print("--- Автоматически найденные производные ---")
for i, d in enumerate(derivatives):
    print(f"По {vars_list[i]}: {d}")

# ВОТ ЗДЕСЬ БЫЛА ОШИБКА: передаем vars_list, а не (x, y)
f_num = sp.lambdify(vars_list, formula, 'numpy')
grad_num = sp.lambdify(vars_list, derivatives, 'numpy')

# ==============================================================================
# БЛОК 3: ПАРАМЕТРЫ ЗАПУСКА
# ==============================================================================
alpha = 0.01        # Шаг
steps = 2000        # Количество итераций
start_point = [0.0, 0.0, 0.0]  # Стартовая точка (3 числа!)

# ==============================================================================
# БЛОК 4: ДВИГАТЕЛЬ (НЕ ТРОГАТЬ)
# ==============================================================================
def gradient_descent(grad_func, start_pt, lr, num_steps):
    point = np.array(start_pt, dtype=float)
    history_points = [point.copy()]
    history_loss = [f_num(*point)]
    
    for i in range(num_steps):
        # Звездочка * распаковывает массив [0.0, 0.0, 0.0] в аргументы x, y, z
        g = np.array(grad_func(*point)) 
        point = point - lr * g
        history_points.append(point.copy())
        history_loss.append(f_num(*point))
        
    return np.array(history_points), np.array(history_loss)

# ==============================================================================
# БЛОК 5: ЗАПУСК И ВЫВОД
# ==============================================================================
history_points, history_loss = gradient_descent(grad_num, start_point, alpha, steps)
final_point = history_points[-1]
final_grad = grad_num(*final_point)

print("\n--- РЕЗУЛЬТАТЫ ОПТИМИЗАЦИИ ---")
print(f"Точка минимума: x = {final_point[0]:.4f}, y = {final_point[1]:.4f}, z = {final_point[2]:.4f}")
print(f"Значение функции: {history_loss[-1]:.4f}")
print(f"Градиент в точке: [{final_grad[0]:.2e}, {final_grad[1]:.2e}, {final_grad[2]:.2e}]")

# ==============================================================================
# БЛОК 6: ВИЗУАЛИЗАЦИЯ
# ==============================================================================
plt.figure(figsize=(8, 4))
plt.plot(history_loss, color='blue', linewidth=2)
plt.title('Как уменьшается ошибка (Loss)')
plt.xlabel('Шаг')
plt.ylabel('f(x,y,z)')
plt.grid(True)
plt.show()
```

### Полное исследование функций на экстремумы + МАТРИЦА ГЕССЕ ###

```python
import sympy as sp

# ==========================================
# 1. ВВОД ПЕРЕМЕННЫХ И ФУНКЦИИ
# ==========================================
x, y = sp.symbols('x y')

# Задаем функцию (для теста взято седло, вы можете поменять на e^x + 5y^3x)
f = 4*x**2 - 6*x*y - 34*x + 5*y**2 + 42*y + 7
# f = sp.exp(x) + 5 * (y**3) * x  # <--- Раскомментируйте это, если хотите решить задачу с картинки

print(f"=== ИССЛЕДОВАНИЕ ФУНКЦИИ: f(x, y) = {f} ===")

# ==========================================
# 2. НАХОДИМ ГРАДИЕНТ (первые производные)
# ==========================================
print("\n[1] ШАГ 1: Ищем градиент (первые производные).")
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)
print(f"    f'_x = {df_dx}")
print(f"    f'_y = {df_dy}")

# ==========================================
# 3. ИЩЕМ КРИТИЧЕСКИЕ ТОЧКИ (где градиент = 0)
# ==========================================
print("\n[2] ШАГ 2: Приравниваем градиент к нулю и ищем точки (решаем систему).")
# Решаем систему уравнений df_dx = 0 и df_dy = 0
critical_points = sp.solve([df_dx, df_dy], [x, y], dict=True)

if not critical_points:
    print("    Критических точек нет (или SymPy не смог их найти аналитически).")
else:
    print(f"    Найдено точек: {len(critical_points)}")
    for pt in critical_points:
        print(f"    -> Точка: x = {pt[x]}, y = {pt[y]}")

# ==========================================
# 4. СТРОИМ МАТРИЦУ ГЕССЕ (вторые производные)
# ==========================================
print("\n[3] ШАГ 3: Строим матрицу Гессе (вторые производные).")
f_xx = sp.diff(f, x, 2)
f_yy = sp.diff(f, y, 2)
f_xy = sp.diff(f, x, y)

H = sp.Matrix([
    [f_xx, f_xy],
    [f_xy, f_yy] # Помним теорему Шварца: f_xy = f_yx
])
print("    Матрица Гессе (в общем виде):")
sp.pprint(H)

# ==========================================
# 5. АНАЛИЗ КАЖДОЙ ТОЧКИ (Мин, Макс или Седло)
# ==========================================
print("\n[4] ШАГ 4: Анализируем точки с помощью Матрицы Гессе.")

if critical_points:
    for pt in critical_points:
        # Подставляем координаты точки в матрицу Гессе
        H_at_pt = H.subs({x: pt[x], y: pt[y]})
        
        # Считаем определитель (детерминант) матрицы в этой точке
        det_H = H_at_pt.det()
        # Считаем значение f''_xx в этой точке
        val_f_xx = H_at_pt[0, 0] 
        
        print(f"\n    --- Анализ точки (x={pt[x]}, y={pt[y]}) ---")
        print(f"    Матрица Гессе в этой точке: {H_at_pt.tolist()}")
        print(f"    Определитель (Δ) = {det_H}")
        print(f"    Вторая производная по x (f''_xx) = {val_f_xx}")
        
        # ЛОГИКА ОПРЕДЕЛЕНИЯ (Тест вторых производных)
        if det_H > 0 and val_f_xx > 0:
            print("    ВЕРДИКТ: ЛОКАЛЬНЫЙ МИНИМУМ (Яма, график изгибается вверх со всех сторон).")
        elif det_H > 0 and val_f_xx < 0:
            print("    ВЕРДИКТ: ЛОКАЛЬНЫЙ МАКСИМУМ (Гора, график изгибается вниз со всех сторон).")
        elif det_H < 0:
            print("    ВЕРДИКТ: СЕДЛОВАЯ ТОЧКА (Ловушка! В одном направлении яма, в другом гора).")
        else:
            print("    ВЕРДИКТ: ТРЕБУЕТСЯ ДОП. ПРОВЕРКА (Определитель = 0, матрица вырождена).")
```

```python
import sympy as sp

# ==============================================================================
# БЛОК 1: ЗДЕСЬ ВЫ ЗАДАЕТЕ ПЕРЕМЕННЫЕ
# ==============================================================================
# Если у вас 1 переменная: оставьте только x
# Если у вас 2 переменные: оставьте x, y
# Если у вас 3 переменные: оставьте x, y, z
x, y = sp.symbols('x y') 

# ==============================================================================
# БЛОК 2: ЗДЕСЬ ВЫ ЗАДАЕТЕ ФУНКЦИЮ (МЕНЯЙТЕ ЭТУ СТРОЧКУ)
# ==============================================================================
# Пример 1: Функция для задачи 8.1 (2 переменные)
f = 2*x**2 - 4*x*y + y**4 + 2

# Пример 2 (если 3 переменные, закомментируйте верхнюю и раскомментируйте эту):
# f = x**3 - 2*x**2 + y**2 + z**2 - 2*x*y + x*z - y*z + 3*z

# Пример 3 (если 1 переменная):
# f = x**3 - 6*x**2 + 2

print(f"=== ИССЛЕДОВАНИЕ ФУНКЦИИ: f = {f} ===")

# ==============================================================================
# БЛОК 3: АВТОМАТИЧЕСКИЙ РАСЧЕТ (НЕ ТРОГАТЬ)
# ==============================================================================
# Получаем список всех переменных, которые вы задали в Блоке 1
vars_list = list(f.free_symbols)

# 1. Находим первые производные (Градиент)
gradient = [sp.diff(f, var) for var in vars_list]
print("\n[1] ГРАДИЕНТ (первые производные):")
for var, g in zip(vars_list, gradient):
    print(f"  f'_{var} = {g}")

# 2. Ищем критические точки (приравниваем градиент к нулю)
print("\n[2] КРИТИЧЕСКИЕ ТОЧКИ (где градиент = 0):")
critical_points = sp.solve(gradient, vars_list, dict=True)

if not critical_points:
    print("  Точек, где все производные равны нулю, не найдено (или SymPy не смог их найти).")
else:
    for pt in critical_points:
        print(f"  Точка: {pt}")

# 3. Строим Матрицу Гессе (вторые производные)
# (Работает автоматически для любого числа переменных)
H = sp.hessian(f, vars_list)
print("\n[3] МАТРИЦА ГЕССЕ (вторые производные):")
sp.pprint(H)

# 4. Анализ каждой критической точки (Мин, Макс, Седло)
print("\n[4] АНАЛИЗ ТОЧЕК (Вердикт):")
if not critical_points:
    print("  Нет точек для анализа.")
else:
    for pt in critical_points:
        # Подставляем точку в Гессиан
        H_at_pt = H.subs(pt)
        
        # Считаем определитель (Δ)
        det_H = H_at_pt.det()
        # Считаем первую вторую производную (верхний левый угол)
        f_xx = H_at_pt[0, 0] 
        
        print(f"\n  --- Точка {pt} ---")
        print(f"  Определитель (Δ) = {det_H}")
        print(f"  f''_xx = {f_xx}")
        
        # Проверка для 1D (одна переменная)
        if len(vars_list) == 1:
            if f_xx > 0: print("  ВЕРДИКТ: МИНИМУМ (яма)")
            elif f_xx < 0: print("  ВЕРДИКТ: МАКСИМУМ (гора)")
            else: print("  ВЕРДИКТ: ТРЕБУЕТ ДОП. ПРОВЕРКИ")

        # Проверка для 2D и более (многомерный случай)
        else:
            if det_H > 0 and f_xx > 0:
                print("  ВЕРДИКТ: ЛОКАЛЬНЫЙ МИНИМУМ (яма)")
            elif det_H > 0 and f_xx < 0:
                print("  ВЕРДИКТ: ЛОКАЛЬНЫЙ МАКСИМУМ (гора)")
            elif det_H < 0:
                print("  ВЕРДИКТ: СЕДЛОВАЯ ТОЧКА (ловушка!)")
            else:
                print("  ВЕРДИКТ: ТРЕБУЕТ ДОП. ПРОВЕРКИ (Δ = 0)")


# ==============================================================================
# БЛОК 4: ЕСЛИ ВАМ НУЖНО ПРОСТО РЕШИТЬ СИСТЕМУ УРАВНЕНИЙ (опционально)
# ==============================================================================
# Если в задаче просят "Найти точки пересечения" или просто "Решить систему",
# а не искать экстремумы, используйте этот блок. 
# (Здесь я взял вашу систему с окружностью и гиперболой)
print("\n" + "="*50)
print("РЕШЕНИЕ СИСТЕМЫ УРАВНЕНИЙ (отдельно от экстремумов):")
eq1 = sp.Eq(3*x**2 + 3*y**2 - 15, 0)
eq2 = sp.Eq(6*x*y - 12, 0)
system_solution = sp.solve([eq1, eq2], [x, y])
print(f"  Найдено {len(system_solution)} решений:")
for sol in system_solution:
    print(f"  {sol}")
```

```python
import sympy as sp

# ==============================================================================
# БЛОК 1: ОБЪЯВЛЕНИЕ ПЕРЕМЕННЫХ
# ==============================================================================
# Если в задаче другие буквы (например, a, b), поменяйте их здесь и в функциях ниже.
x, y, lam = sp.symbols('x y lambda')

# ==============================================================================
# БЛОК 2: ВВОД ДАННЫХ ЗАДАЧИ (МЕНЯЙТЕ ТОЛЬКО ЗДЕСЬ)
# ==============================================================================
# 1. Впишите вашу ЦЕЛЕВУЮ ФУНКЦИЮ f(x, y)
# (То, что нужно найти: максимум или минимум)
f = 3*x**2 + x + 4*y + 5*y**2   # <--- ЗАМЕНИТЕ НА СВОЮ ФУНКЦИЮ

# 2. Впишите ваше ОГРАНИЧЕНИЕ g(x, y) = 0
# (Обязательно перенесите всё в левую часть, чтобы справа был 0)
g = x + y - 200     # <--- ЗАМЕНИТЕ НА СВОЕ ОГРАНИЧЕНИЕ

# ==============================================================================
# БЛОК 3: ФУНКЦИЯ ЛАГРАНЖА И ПРОИЗВОДНЫЕ
# ==============================================================================
# Строим функцию Лагранжа: L = f(x,y) - lambda * g(x,y)
L = f - lam * g

# Берем частные производные по каждой переменной
dL_dx = sp.diff(L, x)       # Производная по x
dL_dy = sp.diff(L, y)       # Производная по y
dL_dlam = sp.diff(L, lam)   # Производная по lambda

# ==============================================================================
# БЛОК 4: РЕШЕНИЕ СИСТЕМЫ УРАВНЕНИЙ
# ==============================================================================
# Приравниваем все три производные к нулю и ищем x, y, lam
# dict=True означает, что результат будет в виде словаря для удобства
solutions = sp.solve([dL_dx, dL_dy, dL_dlam], [x, y, lam], dict=True)

# ==============================================================================
# БЛОК 5: ВЫВОД РЕЗУЛЬТАТОВ
# ==============================================================================
print("="*50)
print("РЕЗУЛЬТАТЫ ОПТИМИЗАЦИИ")
print("="*50)

# Проверяем, есть ли вообще решения (иногда их не бывает)
if len(solutions) == 0:
    print("Решений не найдено. Проверьте функции.")
else:
    # Проходимся по всем найденным решениям (если их несколько)
    for i, sol in enumerate(solutions, 1):
        print(f"\n--- Решение №{i} ---")
        
        # Достаем значения из найденного словаря
        x_val = sol[x]
        y_val = sol[y]
        lam_val = sol[lam]
        
        # Считаем значение целевой функции в этой точке
        f_val = f.subs({x: x_val, y: y_val})
        
        # Выводим все на экран (evalf() переводит дроби в десятичные числа)
        print(f"Точка: x = {x_val.evalf():.4f}, y = {y_val.evalf():.4f}")
        print(f"Значение функции f(x,y) = {f_val.evalf():.4f}")
        print(f"Множитель Лагранжа lambda = {lam_val.evalf():.4f}")
        
        import numpy as np
from scipy.optimize import minimize
```

### Метод Лагранжа численный (универсальный) ДЛЯ ЛЮБОГО КОЛ_ВА ПЕРЕМЕННЫХ ###

```python
import sympy as sp

# ==============================================================================
# БЛОК 1: НАСТРОЙКА КОЛИЧЕСТВА ПЕРЕМЕННЫХ И НАЛИЧИЯ НЕРАВЕНСТВ
# ==============================================================================
# Впишите названия ваших переменных через пробел.
# Для 2 переменных: 'x y'. Для 3 переменных: 'x y z'. Для 4: 'x y z w'.
vars_str = 'x y' 

# Создаем переменные автоматически
variables = sp.symbols(vars_str)
if not isinstance(variables, tuple):
    variables = (variables,)

# ЕСТЬ ЛИ У ВАС ОГРАНИЧЕНИЯ-НЕРАВЕНСТВА?
# Поставьте True, если в задаче есть условия вида g(x) >= 0 или g(x) <= 0.
# Поставьте False, если у вас ТОЛЬКО равенства g(x) = 0.
use_ineq = True 

# Объявляем множители Лагранжа и дополнительную переменную (для неравенств)
# (lambda1 - для первого ограничения, lambda2 - для второго)
lam1, lam2 = sp.symbols('lambda1 lambda2')
s1 = sp.symbols('s1') # Дополнительная переменная для неравенства

# ==============================================================================
# БЛОК 2: ВВОД ВАШИХ ФУНКЦИЙ (МЕНЯЙТЕ ТОЛЬКО ЗДЕСЬ)
# ==============================================================================
# 1. ЦЕЛЕВАЯ ФУНКЦИЯ f(x, y, z...)
# Используйте variables[0], variables[1] и т.д. для обозначения x, y, z.
# Пример для 3 переменных (объем коробки):
f = - variables[0] * variables[1] 

# 2. ОГРАНИЧЕНИЕ-РАВЕНСТВО g1(x) = 0 (Всегда есть)
# Пример: сумма сторон равна 10.
g1 = sum(variables) - 10

# 3. ОГРАНИЧЕНИЕ-НЕРАВЕНСТВО g2(x) >= 0 (Если use_ineq = True)
# Пример: первая переменная должна быть >= 2.
# Записываем как: -x1 + 2 + s1^2 = 0 (чтобы s1^2 "съело" минус)
g2 = -variables[0] + 6 + s1**2

# ==============================================================================
# БЛОК 3: АВТОМАТИЧЕСКИЙ РАСЧЕТ (НЕ МЕНЯТЬ НИЧЕГО ЗДЕСЬ)
# ==============================================================================
# Строим функцию Лагранжа
L = f - lam1 * g1
if use_ineq:
    L = L - lam2 * g2 # Добавляем второе ограничение только если оно есть

# Создаем пустые списки для уравнений и переменных
equations = []
all_symbols = list(variables) + [lam1]

# 1. Берем производные по всем обычным переменным (x, y, z...)
for var in variables:
    equations.append(sp.diff(L, var))

# 2. Берем производную по первому множителю Лагранжа (всегда)
equations.append(sp.diff(L, lam1))

# 3. Если есть неравенство, добавляем производные по s1 и второму множителю
if use_ineq:
    equations.append(sp.diff(L, s1))
    equations.append(sp.diff(L, lam2))
    all_symbols.append(s1)
    all_symbols.append(lam2)

# ==============================================================================
# БЛОК 4: РЕШЕНИЕ И ВЫВОД РЕЗУЛЬТАТОВ
# ==============================================================================
print(f"--- Анализ задачи для переменных: {variables} ---")
print(f"Неравенства включены: {use_ineq}")
print(f"Количество уравнений в системе: {len(equations)}")

# Решаем систему
solutions = sp.solve(equations, all_symbols, dict=True)

if len(solutions) == 0:
    print("\nРешений не найдено. Возможно, система не имеет точного решения или его нужно искать численно.")
else:
    print(f"\nНайдено решений: {len(solutions)}")
    
    for i, sol in enumerate(solutions, 1):
        # Проверяем, что решение действительное (без мнимых чисел 'i')
        is_real = all(sol[s].is_real for s in all_symbols)
        
        if is_real:
            print(f"\n--- Решение №{i} ---")
            # Выводим значения переменных
            for var in variables:
                print(f"{var} = {sol[var].evalf():.4f}")
            
            # Выводим значения дополнительных переменных (если есть неравенство)
            if use_ineq:
                print(f"s1 = {sol[s1].evalf():.4f}")
            
            # Считаем значение целевой функции
            # (подставляем найденные значения переменных)
            f_val = f.subs({var: sol[var] for var in variables})
            print(f"Значение функции f = {f_val.evalf():.4f}")
```

### NA Метод Лагранжа с НЕРАВЕНСТВАМИ как в учебнике просто - потом разобрать! ###

```python
# Импортируем все необходимые функции из библиотеки SymPy
from sympy import init_printing, solve, symbols

# Подключаем аккуратный вывод формул
init_printing()

# Создаём переменные
x1, x2, x1_tilde, lambda1, lambda2 = symbols(
    "x_1 x_2 xtilde_2 \lambda_1 \lambda_2", real=True
)

# Пишем систему уравнений для переменных
dLdx1 = -x2 + lambda1 - lambda2
dLdx2 = -x1 + lambda1
dLdx1_tilde = 2 * lambda2 * x1_tilde
dLdlambda1 = x1 + x2 - 10
dLdlambda2 = -x1 + 6 + x1_tilde**2

# Решаем систему уравнений с помощью функции solve
solution = solve(
    [dLdx1, dLdx2, dLdx1_tilde, dLdlambda1, dLdlambda2],
    x1,
    x2,
    x1_tilde,
    lambda1,
    lambda2,
)

print(solution)
```
