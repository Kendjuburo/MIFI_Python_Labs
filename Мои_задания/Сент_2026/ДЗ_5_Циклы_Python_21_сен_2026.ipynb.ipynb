```python
"""Задача 4. Генератор тестовых данных (random)
Перед нагрузочным тестом нужно сгенерировать входные данные: временные пароли, случайную выборку пользователей и перемешанный порядок сценариев.

random.seed(7) в первой строке делает запуск воспроизводимым — не удаляйте её, иначе проверки будут падать через раз (а сами значения при этом остаются «случайными»).

Нужно:

password — строка из 12 символов, каждый выбран из alphabet (повторы допустимы);
chosen_users — 3 разных пользователя из users;
scenarios_shuffled — перемешанная копия списка scenarios (сам scenarios меняться не должен);
rolls — 20 бросков кубика (randint(1, 6));
counts — словарь «грань → сколько раз выпала».
Проверки написаны на свойства результата, а не на конкретные значения: так тестируют случайность в реальных проектах."""

import random

random.seed(7)

alphabet = "abcdefghijklmnopqrstuvwxyz0123456789"
users = ["anna", "boris", "clara", "dmitry", "elena", "fedor"]
scenarios = ["login", "search", "checkout", "logout"]

random.seed(7)
import random

"""password — строка из 12 символов, каждый выбран из alphabet (повторы допустимы)"""

password = ""

for word_1 in alphabet:
  password += random.choice(alphabet)
  if len(password) > 12:
    break
print(password)

for _ in range(12):
  password += random.choice(alphabet)
print(password)

"""Или таким способом"""

password = "".join(random.choices(alphabet, k=12))
print(password)

"""chosen_users — 3 разных пользователя из users"""

chosen_users = random.sample(users, 3)
print(chosen_users)

"""scenarios_shuffled — перемешанная копия списка scenarios (сам scenarios меняться не должен)"""

scenarios_shuffled = scenarios.copy()
random.shuffle(scenarios_shuffled)
print(scenarios_shuffled)

"""rolls — 20 бросков кубика (randint(1, 6))"""

cube = range(1, 6+1)
print(*cube)

rolls = random.choices(cube, k=20)
print(rolls)

"""2-й способ"""
rolls = []
for _ in range(20):
  rolls += [random.choice(cube)]
print(rolls)

"""counts — словарь «грань → сколько раз выпала»
1. Берем rolls - это какая грань выпадала
делаем count - сколько раз какая грань выпала
обозначаем это переменной и ее переводим в словарь
"""
"""1-е решение"""
counts = {}

for face in range(1, 7):
  counts[face] = rolls.count(face)
print(counts)

"""2-е решение"""

for face_1 in rolls:
  if face_1 in counts:
    counts[face_1] += 1
  else:
    counts[face_1] = 1

print(counts)

"""3-е решение"""

counts = {}

for n in range(1,7):
  counts[n] = rolls.count(n)


print(counts)

print(password)
print(chosen_users)
print(scenarios_shuffled, scenarios)
print(counts)
```
