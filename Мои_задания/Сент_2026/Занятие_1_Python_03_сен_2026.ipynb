```python
"""
Задача 2. Смета на облачный инстанс
"Считаем месячную стоимость виртуальной машины для отчёта заказчику.

Дано:

price_per_hour = 1.37 (руб/час)
hours_per_month = 730
instances = 6
НДС VAT = 0.20
Нужно: строка report вида 6 инстансов × 730 ч = 6 000.60 руб. без НДС, 7 200.72 руб. с НДС

Требования к формату: два знака после запятой, разделитель тысяч — пробел (получите его из , заменой: f"{x:,.2f}".replace(",", " "))."
"""
price_per_hour = 1.37
hours_per_month = 730
instances = 6
VAT = 0.20
price=price_per_hour*hours_per_month*instances
print(price)
price_VAT=price*VAT+price
print(price_VAT)
report=f'{price:,.2f} руб., {price_VAT:,.2f} руб. с НДС'
print(report)
report=report.replace(",", " ")
print(report)
```

```python
'''
Задача 3. Имя артефакта сборки
CI-система должна собрать имя файла по шаблону:

<service>-<major>.<minor>.<patch>-build<НОМЕР_СБОРКИ_5_ЗНАКОВ>.tar.gz

Номер сборки дополняется нулями слева до 5 знаков ({n:05d}).

Дано: service = "billing", major, minor, patch = 2, 14, 3, build = 87
Ожидаем: billing-2.14.3-build00087.tar.gz
'''
service = "billing"
major = 2
minor = 14
patch = 3
build = 87
name = f'{service}-{major}.{minor}.{patch}-build{build:05d}.tar.gz'
print(name)
```

```python
"""
Задача 4. Мониторинг диска
Агент мониторинга прислал сырые значения в байтах (строками — так они пришли из JSON).

Дано:

total_bytes = "512110190592"
used_bytes  = "437260500992"
Нужно:

free_gb — свободно в ГБ (делим на 1024 три раза), округлить до 1 знака (round(x, 1));
used_percent — занято в процентах, округлить до 1 знака;
alert — строка DISK 85.4% USED, FREE 69.7GB.
Формат процента в строке — {x:.1f}%.
"""
total_bytes = "512110190592"
used_bytes  = "437260500992"
total_bytes = int(total_bytes)
used_bytes  = int(used_bytes)
free_bytes = total_bytes - used_bytes
#print(free_bytes)
free_gb = round(free_bytes/1024**3, 1)
#print(free_gb)
used_percent = round(used_bytes / total_bytes*100, 1)
#print(used_percent)
alert = f'DISK {used_percent:.1f}% USED, FREE {free_gb:.1f}GB'
print(alert)
```

```python
"""
Задача 5. Строка конфигурации - 1-е решенние
Из переменных окружения пришла строка подключения. Пока без «взрослого» парсинга — только базовые операции.

Дано: dsn = "postgres://app_user:s3cret@db-01.local:5432/billing"

Нужно (используйте split и срезы, но без регулярных выражений):

user → "app_user"
host → "db-01.local"
port → 5432 (именно число)
database → "billing"
safe_dsn — та же строка, но пароль заменён на ***: postgres://app_user:***@db-01.local:5432/billing
Логировать DSN с паролем — реальный инцидент безопасности, поэтому маскирование обязательно.
"""
dsn = "postgres://app_user:s3cret@db-01.local:5432/billing"
sheme,rest = dsn.split("://",1)
#print(sheme)
#print(rest)
user,rest = rest.split(":",1)
#print(user)
#print(rest)
secret,rest = rest.split("@",1)
#print(secret)
#print(rest)
host,rest = rest.split(":",1)
#print(host)
#print(rest)
port,database = rest.split("/",1)
#print(port)
#print(database)
port = int(port)
safe_dsn = f'{sheme}://{user}:***@{host}:{port}/{database}'
print(safe_dsn)

```

```python
"""
Задача 5. Строка конфигурации - 2-е решение
Из переменных окружения пришла строка подключения. Пока без «взрослого» парсинга — только базовые операции.

Дано: dsn = "postgres://app_user:s3cret@db-01.local:5432/billing"

Нужно (используйте split и срезы, но без регулярных выражений):

user → "app_user"
host → "db-01.local"
port → 5432 (именно число)
database → "billing"
safe_dsn — та же строка, но пароль заменён на ***: postgres://app_user:***@db-01.local:5432/billing
Логировать DSN с паролем — реальный инцидент безопасности, поэтому маскирование обязательно.
"""
dsn = "postgres://app_user:s3cret@db-01.local:5432/billing"
safe_dsn = dsn.replace("s3cret", "***", 1)
print(safe_dsn)
```
