```python
# Внешняя функция для вычисления итоговой стоимости
def calculate_cost(cost, sale):
    # Внутренняя функция для предобработки аргумента sale
    def preprocessing_sale():
        # Объявляем, что используем нелокальную переменную sale
        nonlocal sale
        # Если sale — строка
        if not isinstance(sale, (str, int, float)):
          raise ValueError("Некорректный формат скидки")
        if type(sale) is str:
            # Удаляем из строки '%', приводим к float и делим на 100
            sale = float(sale.replace('%', '')) / 100
        # Если sale — целое число
        elif type(sale) is int:
            # Делим его на 100
            sale = sale / 100

    # Запускаем предобработку, прежде чем вычислить стоимость
    preprocessing_sale()
    # Считаем итоговую стоимость и возвращаем её
    # (стоимость — стоимость * скидка)
    return cost - cost * sale

print(calculate_cost(1000, 1))
```

```python


def register(surname, name, date, middle_name=None, registry=None):
    # Если список не передан — создаём новый
    if registry is None:
        registry = []

    # Вспомогательная функция для проверки корректности даты
    def check_date(day, month, year):
        # Проверка типов (ленивый метод через isinstance)
        if not isinstance(day, int) or not isinstance(month, int) or not isinstance(year, int):
            return False
        # Проверка года (диапазон из твоего кода)
        if not (1900 <= year <= 2022):
            return False
        # Проверка месяца: от 1 до 12
        if not (1 <= month <= 12):
            return False
        # Проверка дня: от 1 до 31
        if not (1 <= day <= 31):
            return False
        # Если все проверки пройдены — дата корректна
        return True

    # Разбор даты через map (ленивый метод)
    day, month, year = map(int, date.split("."))

    # Вызов проверки и выброс исключения при ошибке
    if not check_date(day, month, year):
        raise ValueError("Некорректная дата")

    # Создание кортежа и добавление в список
    registry.append((surname, name, middle_name, day, month, year))
    return registry
```

```python

"""Задание 3. Проверка скобок"""

PAIRS = {")": "(", "]": "[", "}": "{"}
OPEN = set(PAIRS.values())
CLOSE = set(PAIRS.keys())


def find_unbalanced(text):
    """Возвращает индекс первого несбалансированного символа или -1."""
    steak = []

    for k, sign in enumerate(text):
        if sign in OPEN:
            steak.append((k, sign))
        elif sign in CLOSE:
            if not steak:
                return k
            if steak[-1][1] != PAIRS[sign]:
                return k
            steak.pop()

    if steak:
        return steak[0][0]

    return -1


def check_brackets(text):
    """True, если все скобки расставлены верно, иначе False."""
    return find_unbalanced(text) == -1



"""Задание 5. Почти палиндром"""

def normalize(text):
    """Оставляет только буквы и цифры, приводит к нижнему регистру."""
    result = ""
    for let in text:
        if let.isalnum():
            result += let.lower()
    return result


def is_palindrome(text):
    """Проверяет, является ли строка палиндромом."""
    clar = normalize(text)
    return clar == clar[::-1]


def is_almost_palindrome(text):
    """True, если строка станет палиндромом после удаления одного символа."""
    clar = normalize(text)
    left = 0
    right = len(clar) - 1

    while left < right:
        if clar[left] == clar[right]:
            left += 1
            right -= 1
        else:
            without_left = clar[:left] + clar[left+1:]
            without_right = clar[:right] + clar[right+1:]

            if is_palindrome(without_left):
                return True
            if is_palindrome(without_right):
                return True
            return False

    return True

"""Задание 4 Генератор Паролей"""

import random
import string  # модуль string для букв и спецсимволов

# alphabet: + заглавные буквы и спецсимволы
alphabet = string.ascii_letters + string.digits + string.punctuation


def generate_password(length):
    # нужно минимум 3 символа для 3 типов
    if length < 3:
        return "Слишком короткий пароль"

    password = ""


    # по одному обязательному символу каждого типа:
    password += random.choice(string.ascii_letters)  # буква
    password += random.choice(string.digits)          # цифра
    password += random.choice(string.punctuation)     # спецсимвол

    # добиваем остаток длины случайными символами
    for _ in range(length - 3):
        password += random.choice(alphabet)

    # Перемешиваем
    password_list = list(password)
    random.shuffle(password_list)

    return "".join(password_list)

# проверка
print(generate_password(12))

```

```python
def is_empty(stack):
  return len(stack) == 0

def stack_size(stack):
  return len(stack)

def push(stack, item):
  return stack.append(item)

def pop(stack):
  if is_empty(stack):
    return None
  else:
    return stack.pop()

def peek(stack):
  if is_empty(stack):
      return None
  else:
    return stack[-1]


s = []
assert is_empty(s) is True
assert stack_size(s) == 0
assert pop(s) is None
assert peek(s) is None

push(s, 1)
push(s, 2)
push(s, 3)
assert stack_size(s) == 3
assert peek(s) == 3
assert is_empty(s) is False

assert pop(s) == 3
assert pop(s) == 2
assert pop(s) == 1
assert pop(s) is None
assert is_empty(s) is True
```
