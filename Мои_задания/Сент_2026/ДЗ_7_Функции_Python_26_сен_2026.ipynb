```python
"""Напиши функцию add(a, b), которая возвращает сумму двух чисел.

Пример:

python
print(add(2, 3))   # 5"""

def add(*dig):
  result = 0
  for x in dig:
     result += x
  return result
print(add(2, 3))


"""Напиши функцию square(x), которая возвращает квадрат числа.

Пример:

python
print(square(4))   # 16"""

def sq(dig_1, n):
  res_1 = dig_1**n
  return res_1
print(sq(5, 2))


"""Напиши функцию greet(name), которая возвращает строку "Привет, <name>!".

Пример:

python
print(greet("Анна"))   # Привет, Анна!"""

def greet(name):
  return "Привет, {}".format(name)
print(greet("Anna"))


"""Напиши функцию full_name(first, last), которая возвращает строку "first, last"..

Пример:

python
print(full_name("Иван", "Петров"))   # Иван Петров"""

def full_name(first, last):
  # return "<{}>, <{}>".format(first, last)
  return f"<{first}>, <{last}>"
print(full_name("Иван", "Петров"))

"""Напиши функцию is_even(n), которая возвращает True, если число чётное, и False — если нет.

Подсказка: чётное число делится на 2 без остатка. В Python остаток от деления — это %.

Пример:

python
print(is_even(4))   # True
print(is_even(5))   # False"""

def is_even(n):
  if n % 2 == 0:
    return True
  else:
    return False
print(is_even(5))

"""Напиши функцию max_of_two(a, b), которая возвращает большее из двух чисел.

Пример:

python
print(max_of_two(3, 7))   # 7
print(max_of_two(10, 2))  # 10"""

def max_of_two(a, b):
  return max(a, b)
print(max_of_two(5,7))

"""Напиши функцию sign(x), которая возвращает:

1, если x > 0;

-1, если x < 0;

0, если x == 0."""

def func(x):
  if x > 0:
    return 1
  elif x < 0:
    return -1
  else:
    return 0

print(func(5))

"""Напиши функцию grade(score), которая возвращает оценку по баллам:

score >= 90 → "A";

score >= 75 → "B";

score >= 60 → "C";

иначе → "F"."""

def grade(score):
  if score >= 90:
    return "A"
  elif score >= 75:
    return "B"
  elif score >= 60:
    return "C"
  else:
    return "F"

print(grade(90))

"""Напиши функцию is_leap(year), которая возвращает True, если год високосный, и False — иначе.

Правило високосного года:

год делится на 4 и не делится на 100;

или делится на 400.

Пример:

python
print(is_leap(2024))   # True
print(is_leap(1900))   # False
print(is_leap(2000))   # True"""

def is_leap(year):
  return year % 4 == 0 and year % 100 != 0 or year % 400 == 0


print(is_leap(2024))

"""Напиши функцию discount(price, percent), которая возвращает цену со скидкой.

Пример:

python
print(discount(1000, 10))   # 900.0
print(discount(500, 0))     # 500.0"""

def discount(price, percent):
  return price - price*percent/100
print(discount(1000, 10))

"""
Напиши функцию sum_list(numbers), которая возвращает сумму всех чисел в списке.

Пример:

python
print(sum_list([1, 2, 3]))     # 6
print(sum_list([]))            # 0"""

def sum_list(numbers):
  return sum(numbers)
print(sum_list([1,2,5]))

"""Напиши функцию count_positive(numbers), которая считает, сколько чисел в списке больше нуля.

Пример:

python
print(count_positive([1, -2, 3, 0, 5]))   # 3"""

def count_positive(numbers):
  num_1 = 0
  for x in numbers:
    if x > 0:
      num_1 += 1

  return num_1

print(count_positive([1, -2, 3, 0, 5]))

"""
Напиши функцию max_in_list(numbers), которая возвращает максимальное число в списке.
Если список пуст — верни None.

Пример:

python
print(max_in_list([3, 1, 7, 2]))   # 7
print(max_in_list([]))
"""

# def max_in_list(numbers):
#   if len(numbers) == 0:
#     return None
#   else:
#     return max(numbers)

def max_in_list(numbers):
  return max(numbers) if numbers else None

print(max_in_list([3, 1, 7, 2]))


"""Напиши функцию reverse_list(items), которая возвращает новый список — тот же, но в обратном порядке.

Не используй reverse() и [::-1] — сделай через цикл.

Пример:

python
print(reverse_list([1, 2, 3]))   # [3, 2, 1]"""


def reverse_list(items):
  result = []
  for x in range(1, len(items) +1):
    result.append(items[-x])
  return result
print(reverse_list([1, 2, 3]))

"""Напиши функцию find_index(items, target), которая возвращает индекс первого вхождения target в список.
Если элемента нет — верни -1.

Пример:

python
print(find_index(["a", "b", "c"], "b"))   # 1
print(find_index([1, 2, 3], 5))            # -1"""

def find_index(items, target):

  for i, n in enumerate(items):
    if target == n:
      return i
  return -1

print(find_index(["a", "b", "c"], "b"))


"""
Напиши функцию is_palindrome(text), которая возвращает True, если строка читается одинаково в обе стороны.

Пример:

python
print(is_palindrome("шалаш"))    # True
print(is_palindrome("питон"))    # False"""

def is_palindrome(text):
      return text == text[::-1]
print(is_palindrome("шалаш"))


"""Напиши функцию normalize(text), которая:

оставляет только буквы и цифры;

приводит всё к нижнему регистру.

Затем перепиши is_palindrome так, чтобы она использовала normalize.

Пример:

python
print(normalize("А роза упала на лапу Азора"))
# арозаупаланалапуазора

print(is_palindrome("А роза упала на лапу Азора"))   # True"""

def normalize(text):
  result = ""
  for let in text:
    if let.isalnum():
      result += let.lower()
  return result

def is_palindrome(text):
  clar = normalize(text)
  return clar == clar[::-1]

print(normalize("А роза упала на лапу Азора"))
print(is_palindrome("А роза упала на лапу Азора"))

"""Напиши функцию is_almost_palindrome(text), которая возвращает True, если строка станет палиндромом после удаления не более одного символа.

Идея:

нормализуй строку;

иди двумя указателями навстречу;

как только символы не совпали — попробуй удалить левый и правый, проверь оба варианта;

если дошли до середины — это палиндром.

Подсказка: удалить символ по индексу i — это s[:i] + s[i+1:].

Пример:

python
print(is_almost_palindrome("шалаш"))    # True (уже палиндром)
print(is_almost_palindrome("шаалаш"))   # True (лишняя «а»)
print(is_almost_palindrome("питон"))    # False
"""
def normalize(text):
  result = ""
  for let in text:
    if let.isalnum():
      result += let.lower()
  return result

def is_palindrome(text):
  clar = normalize(text)
  return clar == clar[::-1]


def is_almost_palindrome(text):
  clar = normalize(text)
  left = 0
  right = len(clar) -1

  while left < right:
    if clar[left] == clar[right]:
      left += 1
      right -= 1

    else:
      without_left = clar[0:left] + clar[left + 1:]
      without_right = clar[0:right] + clar[right + 1:]

      if is_palindrome(without_left):
        return True
      if is_palindrome(without_right):
        return True

      return False

  return True
```

```python
"""PAIRS = {")": "(", "]": "[", "}": "{"}
OPEN = set(PAIRS.values())    # {"(", "[", "{"}
CLOSE = set(PAIRS.keys())     # {")", "]", "}"}

find_unbalanced(text) — индекс первого несбалансированного символа

check_brackets(text) — True/False"""

"""Напиши функцию find_unbalanced(text), которая возвращает индекс первого проблемного символа, или -1, если всё в порядке.

Что значит «проблемный»:

лишняя/непарная закрывающая → её собственный индекс;

незакрытая открывающая → индекс этой открывающей;

если всё хорошо → -1.

Важно: хранить в стеке не только символ, но и его позицию:

python
stack.append((i, ch))    # кортеж (индекс, символ)
Пример:

python

"""


PAIRS = {")": "(", "]": "[", "}": "{"}
OPEN = set(PAIRS.values())    # {"(", "[", "{"}
CLOSE = set(PAIRS.keys())     # {")", "]", "}"}

""" Алгоритм:
1. разбиваем текст на интекся и значения enumerate(text)
2. создаем stack куда кладем все скобки из текста
3. если определенной скобке в тексте находитс соотвтетвующая - необходимо удалить как последнюю так и соответсвующую (pop - для плследней как для первой неопнятно)
4. если после удаления в steak остается скобка - выносим ее с инелдесом
5. в противном слоучае -1
"""

# def find_unbalanced(text):
#   steak = []
#   for k,sign in enumerate(text):

def find_unbalanced(text):
    steak = []
    for k, sign in enumerate(text):
        if sign in OPEN:
            steak.append((k, sign))
        elif sign in CLOSE:
            if not steak:
                return k
            if steak[-1][1] != PAIRS[sign]:
                return k
            steak.pop()

    if steak:
        return steak[0][0]     # ← ЭТА СТРОКА

    return -1

def check_brackets(text):
  return find_unbalanced(text) == -1

print(check_brackets("([{}])"))
print(check_brackets("([)]"))
print(check_brackets("((a)"))
print(check_brackets(""))
print(check_brackets("a)b"))


assert check_brackets("()") is True
assert check_brackets("[]") is True
assert check_brackets("{}") is True
assert check_brackets("([{}])") is True
assert check_brackets("a(b)[c]{d}") is True
assert check_brackets("") is True
assert check_brackets("def f(x): return [x, {1: (2, 3)}]") is True

assert check_brackets("(") is False
assert check_brackets(")") is False
assert check_brackets("([)]") is False
assert check_brackets("((a)") is False
assert check_brackets("a)b") is False
assert check_brackets("[(])") is False

assert find_unbalanced("([{}])") == -1
assert find_unbalanced("") == -1
assert find_unbalanced("a)b") == 1
assert find_unbalanced("([)]") == 2
assert find_unbalanced("((a)") == 0
assert find_unbalanced("func(a, [b, c)") == 13
```
