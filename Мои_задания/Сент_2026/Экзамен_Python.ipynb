```python
"""Задание 3. Проверка скобок"""

PAIRS = {")": "(", "]": "[", "}": "{"}
OPEN = set(PAIRS.values())
CLOSE = set(PAIRS.keys())


def find_unbalanced(text):
    """Возвращает индекс первого несбалансированного символа или -1."""
    steak = []

    for k, sign in enumerate(text):
        if sign in OPEN:
            steak.append((k, sign))
        elif sign in CLOSE:
            if not steak:
                return k
            if steak[-1][1] != PAIRS[sign]:
                return k
            steak.pop()

    if steak:
        return steak[0][0]

    return -1


def check_brackets(text):
    """True, если все скобки расставлены верно, иначе False."""
    return find_unbalanced(text) == -1

print(find_unbalanced("((((((({{{{{{{[[[[[]]]]]}}}}}}})))))]]))"))
print(find_unbalanced("((((([[[[[{{{{{{]}}}}}}]]]]])))))"))
print(find_unbalanced("((((([[[[[{{{{{{}}}}}}]]]]])))))"))
```

```python
"""Задание 5. Почти палиндром"""

def normalize(text):
    """Оставляет только буквы и цифры, приводит к нижнему регистру."""
    result = ""
    for let in text:
        if let.isalnum():
            result += let.lower()
    return result


def is_palindrome(text):
    """Проверяет, является ли строка палиндромом."""
    clar = normalize(text)
    return clar == clar[::-1]


def is_almost_palindrome(text):
    """True, если строка станет палиндромом после удаления одного символа."""
    clar = normalize(text)
    left = 0
    right = len(clar) - 1

    while left < right:
        if clar[left] == clar[right]:
            left += 1
            right -= 1
        else:
            without_left = clar[:left] + clar[left+1:]
            without_right = clar[:right] + clar[right+1:]

            if is_palindrome(without_left):
                return True
            if is_palindrome(without_right):
                return True
            return False

    return True

# 1. Проверка normalize
print(normalize("Hello, World! 123"))          # helloworld123
print(normalize("А роза упала на лапу Азора"))  # арозаупаланалапуазора

# 2. Проверка is_palindrome
print(is_palindrome("А роза упала на лапу Азора"))  # True
print(is_palindrome("Привет"))                      # False

# 3. Проверка is_almost_palindrome
print(is_almost_palindrome("А роза упала на лапу Азора"))   # True (уже палиндром)
print(is_almost_palindrome("А роза упала на лапу Азора!"))  # True (после удаления '!' станет палиндромом)
print(is_almost_palindrome("abca"))                         # True
print(is_almost_palindrome("abc"))
```

```python
"""Задание 4 Генератор Паролей"""

import random
import string  # модуль string для букв и спецсимволов

# alphabet: + заглавные буквы и спецсимволы
alphabet = string.ascii_letters + string.digits + string.punctuation


def generate_password(length):
    # нужно минимум 3 символа для 3 типов
    if length < 3:
        return "Слишком короткий пароль"

    password = ""


    # по одному обязательному символу каждого типа:
    password += random.choice(string.ascii_letters)  # буква
    password += random.choice(string.digits)          # цифра
    password += random.choice(string.punctuation)     # спецсимвол

    # добиваем остаток длины случайными символами
    for _ in range(length - 3):
        password += random.choice(alphabet)

    # Перемешиваем
    password_list = list(password)
    random.shuffle(password_list)

    return "".join(password_list)

# проверка
print(generate_password(12))


# 1. Обычный пароль длиной 12 символов
print(generate_password(12))

# 2. Минимально допустимая длина — 3 символа
print(generate_password(3))

# 3. Слишком короткая длина — вернётся сообщение об ошибке
print(generate_password(2))

# 4. Длинный пароль
print(generate_password(20))
```
